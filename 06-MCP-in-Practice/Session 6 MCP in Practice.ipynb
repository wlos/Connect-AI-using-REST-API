## 1. MCP Resources and Prompt Design

[Session 5's weather server](../05-Introduction-to-MCP/practice/weather_mcp_server.py) introduced `@mcp.resource()` and `@mcp.prompt()`. In this session, the host uses the resource URI to choose data and fills in a prompt with the audience and city.

| MCP primitive | Who initiates the operation in this example | What the server returns |
|---|---|---|
| Resource | The host selects a URI and reads it. | A catalogue of cities or one city's weather record. |
| Prompt | The learner selects a briefing and the host supplies its arguments. | Rendered messages for the host to review or send to an LLM. |
| Tool | The host executes a tool requested by the model, subject to the host's tool policy. | The result of the call, such as the selected weather sample. |

### Design and read weather resources

In [sample_weather_mcp_server.py](practice/sample_weather_mcp_server.py), both resources use [weather_samples.json](sample_data/weather_samples.json). The fixed URI `weather://cities` lists the available cities as Markdown. The template `weather://sample/{city}` accepts a city segment and returns its record as JSON.

```python
@mcp.resource("weather://cities", mime_type="text/markdown")
def available_cities() -> str:
    """List the cities that have simulated weather records."""
    cities = [f"- {sample['city']} (`@{key}`)" for key, sample in sorted(load_samples().items())]
    return "\n".join(["# Available weather samples", "", *cities])


@mcp.resource("weather://sample/{city}", mime_type="application/json")
def weather_sample(city: str) -> dict:
    """Read the sample for a city through a resource URI."""
    return sample_weather(city)
```

On the client, `list_tools()`, `list_resources()` and `list_resource_templates()` discover different server capabilities. Read the catalogue first, then construct a URI for a city it lists. Try `weather://sample/sydney` to see a different record.

The notebook runs the server in the same process. Run these cells from the session directory.

[FastMCP resources](https://gofastmcp.com/clients/resources)


````python
import json
import sys
from fastmcp import Client

# Make the session's practice modules importable from the notebook.
sys.path.insert(0, "practice")
from sample_weather_mcp_server import mcp as weather_server

async with Client(weather_server) as weather:
    print("Tools:", [tool.name for tool in await weather.list_tools()])
    print("Fixed resources:", [str(item.uri) for item in await weather.list_resources()])
    templates = await weather.list_resource_templates()
    print("URI templates:", [item.uri_template for item in templates])

    # Read the Markdown catalogue, then request one JSON record.
    catalogue, = await weather.read_resource("weather://cities")
    print(catalogue.text)
    sample, = await weather.read_resource("weather://sample/perth")
    perth_weather = json.loads(sample.text)
    print(perth_weather)
````

### Design and use a weather prompt

The `@mcp.prompt()` decorator publishes a named template with arguments. The learner supplies `city` and can override the default `audience`. The template gives the model an ordered task, specifies the output length and tells it how to handle missing data.

```python
@mcp.prompt()
def weather_briefing(city: str, audience: str = "event participants") -> str:
    """Prepare an activity briefing with a clear audience and output format."""
    return (
        f"Write a two-sentence weather briefing for {audience} in {city}. "
        f"\n1. Use the sample supplied by the host or call get_weather(city={city!r}) "
        "if the tool is available. Ask for the sample if neither source is available."
        "\n2. State the temperature in degrees Celsius and the conditions. Label both as simulated."
        "\n3. Suggest one practical preparation for the audience."
    )
```

The client discovers prompt names and their argument metadata with `list_prompts()`. It then calls `get_prompt()` with the learner's values to retrieve the rendered messages. The host can show these messages for review or send them to a model along with the sample it read earlier.

[FastMCP prompts](https://gofastmcp.com/clients/prompts)


````python
async with Client(weather_server) as weather:
    for prompt in await weather.list_prompts():
        print(prompt.name, "-", prompt.description)
        print([(arg.name, arg.required) for arg in (prompt.arguments or [])])

    # Supply both arguments to tailor the rendered prompt.
    briefing = await weather.get_prompt(
        "weather_briefing",
        arguments={"city": "Perth", "audience": "photography students"},
    )

for message in briefing.messages:
    print(message.role, message.content.text)
print("Weather context:", json.dumps(perth_weather, ensure_ascii=False))
````

### Try the terminal commands

The [weather resource client](practice/weather_resource_client.py) translates short commands into MCP requests. The shortcuts belong to this client. A client could provide the same operations through a menu or a web form.

| Command | Request |
|---|---|
| `@cities` | Read the fixed resource at `weather://cities`. |
| `@perth` | Fill the city segment and read `weather://sample/perth`. |

| `/prompt weather_briefing city=Perth` | Render the briefing using the default audience. |
| `/prompt weather_briefing city=Perth audience="photography students"` | Render the briefing for a specific audience. Quotes keep the value together. |

Run the same command handler in the notebook before trying the terminal loop.


````python
from weather_resource_client import handle_command

async with Client(weather_server) as weather:
    # Keep one client open while issuing several commands.
    for query in (
        "@cities",
        "@perth",
        "/prompts",
        '/prompt weather_briefing city=Perth audience="photography students"',
    ):
        print(f"MCP> {query}")
        print(await handle_command(weather, query))
````

## 2. Connect Multiple MCP Servers

The weather server provides `get_weather(city)`. The [Filesystem server](practice/filesystem_mcp_server.py) provides `read_event_plan()`, which reads the bundled [event plan](sample_data/event-plan.md). The Filesystem server chooses the file, so its callers do not supply a path.

### Connect and inspect each server

Store the server entries in [mcp_servers.json](practice/mcp_servers.json). Each name under `mcpServers` identifies a server. The app uses it as a prefix for tools shown to the model.

```json
{
  "mcpServers": {
    "weather": {
      "args": ["sample_weather_mcp_server.py", "--transport", "stdio"],
      "url_env": "MCP_SERVER_URL",
      "allowed_tools": ["get_weather"]
    },
    "files": {
      "args": ["filesystem_mcp_server.py"],
      "allowed_tools": ["read_event_plan"]
    }
  }
}
```

- `args` gives the script filename and arguments. The app starts local servers with the current Python interpreter and connects over stdio.
- `url_env` names an optional environment variable containing a Streamable HTTP URL. When `MCP_SERVER_URL` is set, the weather client connects to that endpoint. The Filesystem client still uses stdio.
- `allowed_tools` lists the tools the chatbot may pass to the model.

[multi_server_chatbot.py](practice/multi_server_chatbot.py) reads the JSON and creates one client per entry. `connect_to_servers()` loops over those clients and calls [Session 5's `connect_to_server()`](../05-Introduction-to-MCP/practice/mcp_chatbot.py) for each. The single-server method opens one connection and discovers its tools. An `AsyncExitStack` keeps all connections open until the caller finishes and closes any that already opened if a later connection fails.

Run the next cell to inspect the configured servers and call one tool from each. Leave `MCP_SERVER_URL` unset for the local stdio exercise. Tool calls in this cell need no LLM API key.


````python
from multi_server_chatbot import MultiServerChatbot

chatbot = MultiServerChatbot()
print("Configured servers:", list(chatbot.servers))

async with chatbot.connect_to_servers():
    print("Routed tools:", list(chatbot.routes))

    # Use the prefixed names the model sees to call the owning server.
    print(await chatbot.execute_tool("weather_get_weather", {"city": "Perth"}))
    print(await chatbot.execute_tool("files_read_event_plan", {}))
````

### Extend the Session 5 chatbot

The Session 6 chatbot inherits the Session 5 chat loop for model requests, streamed replies and conversation history. Once each client is open, `connect_to_servers()` checks its discovered tool names against the JSON allowlist. It prefixes each allowed tool's name and records the owning client and original name.

| Name sent to the model | MCP client | Original tool name |
|---|---|---|
| `weather_get_weather` | Weather | `get_weather` |
| `files_read_event_plan` | Filesystem | `read_event_plan` |

When the model requests a tool, `execute_tool()` looks up the prefixed name and sends the original name and arguments to the owning client. The inherited chat loop passes the result back under the model's original tool-call ID. The prefix keeps tools with the same original name distinct across servers.

## 3. Deploy a Remote MCP Server

The [weather MCP server](practice/sample_weather_mcp_server.py) exposes simulated weather over Streamable HTTP. Test it on your computer first, then deploy the same server to Render. It needs no API keys.

### Test the MCP server locally

From the session directory, start the server in a terminal.

```bash
python practice/sample_weather_mcp_server.py --host 127.0.0.1 --port 8000
```

Open `http://127.0.0.1:8000/health` and check for `OK`. Then set `CHECK_HTTP` to `True` and run the cell below. The MCP client should find `get_weather` and read the Perth sample. Leave the local server running for section 4.


````python
from fastmcp import Client

CHECK_HTTP = False  # Enable after starting the weather server.
mcp_url = "http://127.0.0.1:8000/mcp"

if CHECK_HTTP:
    async with Client(mcp_url, timeout=30) as weather:
        print("Tools:", [tool.name for tool in await weather.list_tools()])
        sample, = await weather.read_resource("weather://sample/perth")
        print(sample.text)
````

### Create the Render weather service

1. Push the course exercise files to a branch in your own repository that Render can access.
2. In the Render Dashboard, choose **New → Web Service**. Select your repository and the branch containing the exercise files.
3. Enter the settings below and deploy. You can set the health check path under the service's **Settings → Health Checks**.

| Setting | Value |
|---|---|
| Name | A unique name for your weather service |
| Language | Python 3 |
| Root Directory | `06-MCP-in-Practice` |
| Build Command | `python -m pip install -r practice/requirements.txt` |
| Start Command | `python practice/sample_weather_mcp_server.py --host 0.0.0.0 --port "$PORT"` |
| Health Check Path | `/health` |

Render runs the commands from the selected root directory. The server listens on `0.0.0.0` and uses the `PORT` that Render provides.

[Render Web Services](https://render.com/docs/web-services) · [Render health checks](https://render.com/docs/health-checks)

### Verify the deployed MCP endpoint

When the deploy succeeds, copy the weather service's HTTPS URL from its Render dashboard. Open that URL with `/health` appended and check for `OK`. In the test cell above, replace `mcp_url` with the weather service's HTTPS URL ending in `/mcp` and run the cell again. Expect the same tool and Perth sample. Keep the full HTTPS `/mcp` URL for the Render deployment in section 4.


## 4. Deploy the Web Chat

[The FastAPI web chat](practice/web_chat.py) connects to the weather MCP server over Streamable HTTP and runs the Filesystem server as a subprocess. Test the browser, FastAPI and both MCP servers locally before deploying the web chat as a second Render service.

| Connection | Interface |
|---|---|
| Browser → web chat service | Same-origin `POST /chat` |
| Web chat service → LLM provider | HTTPS chat completions |
| Web chat service → weather service | Streamable HTTP at the weather server's `/mcp` URL |
| Web chat service → Filesystem server | stdio to the bundled subprocess |

### Test the web chat locally

Keep the local weather server from section 3 running. In a new terminal, set the provider variables from section 2 and set `MCP_SERVER_URL` to `http://127.0.0.1:8000/mcp`. From the session directory, start FastAPI.

